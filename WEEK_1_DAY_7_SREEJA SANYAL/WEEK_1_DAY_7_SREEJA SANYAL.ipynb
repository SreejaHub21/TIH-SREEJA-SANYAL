```python
import os
import pandas as pd
base_path = "/content/drive/MyDrive/BATTERY_SOC_PROJECT/SOC DATA SET/Dataset_Li-ion"
temperature_folders = ["0degC","10degC","25degC","40degC","n10degC","n20degC"]
print("Dataset path:")
print(base_path)
print("\nChecking temperature folders")
for folder in temperature_folders:
    folder_path = os.path.join(base_path, folder)
    if os.path.exists(folder_path):
        print(folder, "- Found")
    else:
        print(folder, "- Not Found")
```

```python
print("\nCSV files in each folder")
total_files = 0
for folder in temperature_folders:
    folder_path = os.path.join(base_path, folder)
    files = os.listdir(folder_path)
    count = 0
    for file in files:
        if file.endswith(".csv"):
            count = count + 1
    print(folder, ":", count, "CSV files")
    total_files = total_files + count
print("\nTotal CSV files:", total_files)
```

```python
file_path = os.path.join(
    base_path,
    "0degC",
    "589_Charge1.csv")
print("First 35 lines of the original CSV")
with open(file_path, "r") as file:
    for i in range(35):
        line = file.readline()
        print(i + 1, ":", line.strip())
```

```python
print("\nInspecting all CSV files")
results = []
for folder in temperature_folders:
    folder_path = os.path.join(base_path, folder)
    for file in os.listdir(folder_path):
        if file.endswith(".csv"):
            file_path = os.path.join(folder_path, file)
            data = pd.read_csv(
                file_path,
                skiprows=28,
                low_memory=False)
            data = data.iloc[1:].copy()
            rows = data.shape[0]
            columns = data.shape[1]
            missing = data.isnull().sum().sum()
            duplicates = data.duplicated().sum()
            results.append([
                folder,
                file,
                rows,
                columns,
                missing,
                duplicates])
print("Inspection completed.")
print("Files inspected:", len(results))
```

```python
inspection = pd.DataFrame(
    results,
    columns=["Temperature","File","Rows","Columns","Missing Values","Duplicate Rows"])
display(inspection)
```

```python
print("Dataset Summary")
print("Total CSV files:", len(inspection))
print("\nFiles in each temperature folder:")
for folder in temperature_folders:
    count = 0
    for value in inspection["Temperature"]:
        if value == folder:
            count = count + 1
    print(folder, ":", count)
```

```python
print("\nMissing Values:-")
total_missing = inspection["Missing Values"].sum()
print("Total missing values:", total_missing)
print("\nDuplicate Rows:-")
total_duplicates = inspection["Duplicate Rows"].sum()
print("Total duplicate rows:", total_duplicates)
```

```python
print("Checking for SOC or target columns")
soc_found = False
for folder in temperature_folders:
    folder_path = os.path.join(base_path, folder)
    for file in os.listdir(folder_path):
        if file.endswith(".csv"):
            file_path = os.path.join(folder_path, file)
            data = pd.read_csv(
                file_path,
                skiprows=28,
                low_memory=False)
            data = data.iloc[1:].copy()
            for column in data.columns:
                if "SOC" in str(column).upper():
                    print("SOC column found:", column)
                    soc_found = True
if soc_found == False:
    print("No direct SOC column was found.")
```

```python
print("Invalid numerical data check")
columns_to_check = ["Voltage","Current","Temperature","Capacity","WhAccu"]
invalid_total = 0
for folder in temperature_folders:
    folder_path = os.path.join(base_path, folder)
    for file in os.listdir(folder_path):
        if file.endswith(".csv"):
            file_path = os.path.join(folder_path, file)
            data = pd.read_csv(
                file_path,
                skiprows=28,
                low_memory=False
            )
            data = data.iloc[1:].copy()
            for column in columns_to_check:
                if column in data.columns:
                    original = data[column]
                    converted = pd.to_numeric(
                        original,
                        errors="coerce"
                    )
                    invalid = (
                        original.notna() &
                        converted.isna()
                    ).sum()
                    if invalid > 0:
                        print(
                            folder,
                            file,
                            column,
                            ":", invalid,
                            "invalid values"
                        )
                        invalid_total = invalid_total + invalid
print("\nTotal invalid values:", invalid_total)
```

```python
print("Simple consistency check")
problem_files = 0
for folder in temperature_folders:
    folder_path = os.path.join(base_path, folder)
    for file in os.listdir(folder_path):
        if file.endswith(".csv"):
            file_path = os.path.join(folder_path, file)
            data = pd.read_csv(
                file_path,
                skiprows=28,
                low_memory=False
            )
            data = data.iloc[1:].copy()
            if "Voltage" in data.columns:
                voltage = pd.to_numeric(
                    data["Voltage"],
                    errors="coerce"
                )
                if (voltage <= 0).any():
                    print(
                        "Non-positive voltage found:",
                        folder,
                        file
                    )
                    problem_files = problem_files + 1
print("\nFiles with non-positive voltage:", problem_files)
```

```python
summary = []
for folder in temperature_folders:
    folder_data = inspection[
        inspection["Temperature"] == folder
    ]
    number_of_files = len(folder_data)
    total_rows = folder_data["Rows"].sum()
    total_missing = folder_data["Missing Values"].sum()
    total_duplicates = folder_data["Duplicate Rows"].sum()
    summary.append([
        folder,
        number_of_files,
        total_rows,
        total_missing,
        total_duplicates
    ])
temperature_summary = pd.DataFrame(
    summary,
    columns=["Temperature","Number of Files","Total Rows","Total Missing Values","Total Duplicate Rows"])
display(temperature_summary)
summary_file = os.path.join(
    report_folder,
    "Day7_Temperature_Summary.csv"
)
temperature_summary.to_csv(
    summary_file,
    index=False
)
print("\nTemperature summary saved:")
print(summary_file)
```
